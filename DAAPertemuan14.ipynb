```python
import networkx as nx
import matplotlib.pyplot as plt

fraud = [1, 4, 5, 7, 10]
non_fraud = [2, 3, 6, 8, 9]
edges = [(1,2), (1,7), (2,3), (2,6), (3,4), (3,6), (5,6), (6,7), (7,8), (7,10), (8,9)]

G = nx.Graph()
G.add_nodes_from(fraud, status='F')
G.add_nodes_from(non_fraud, status='NF')
G.add_edges_from(edges)

print(f"{'Node':<5}|{'Stat':<5}|{'Deg':<5}|{'F-Neigh':<8}|{'Prob'}")
for n in sorted(G.nodes()):
    nbrs = list(G.neighbors(n))
    f_nbrs = sum(1 for x in nbrs if G.nodes[x]['status']=='F')
    prob = f_nbrs / len(nbrs) if nbrs else 0
    print(f"{n:<5}|{G.nodes[n]['status']:<5}|{len(nbrs):<5}|{f_nbrs:<8}|{prob:.2f}")

colors = ['red' if G.nodes[n]['status'] == 'F' else 'green' for n in G.nodes()]
nx.draw(G, with_labels=True, node_color=colors, font_color='white', font_weight='bold')
plt.show()
```
